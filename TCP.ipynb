```python
from scapy.all import *

dst_ip = "127.0.0.1"
dst_port = 8888

syn = IP(dst=dst_ip) / TCP(dport=dst_port, flags="S")
syn_ack = sr1(syn, timeout=5)

ack = IP(dst=dst_ip) / TCP(dport=dst_port,
                            sport=syn_ack[TCP].dport,
                            seq=syn_ack[TCP].ack,
                            ack=syn_ack[TCP].seq + 1,
                            flags="A")
send(ack, verbose=False)

print("Handshake completado")
```
