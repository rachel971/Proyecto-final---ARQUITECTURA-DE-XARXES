```python
from scapy.all import *

# Crear paquete ICMP Echo Request (ping)
paquete = IP(dst="8.8.8.8") / ICMP()

# Enviar y esperar respuesta
respuesta = sr1(paquete, timeout=5)

if respuesta:
  print("Respuesta recibida:")
  respuesta.show()
else:
  print("No hubo respuesta")
```
