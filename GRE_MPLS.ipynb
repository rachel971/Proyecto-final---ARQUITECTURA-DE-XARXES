```python
from scapy.all import *

paquete_interno = IP(dst="8.8.8.8") / ICMP()
paquete_gre = IP(dst="127.0.0.1") / GRE(proto=0x0800) / paquete_interno
paquete_gre.show()
send(paquete_gre, verbose=False)
```

```python
from scapy.all import *
from scapy.contrib.mpls import MPLS

paquete_mpls = IP(dst="8.8.8.8") / MPLS(label=100, ttl=64) / ICMP()
paquete_mpls.show()
send(paquete_mpls, verbose=False)
```
