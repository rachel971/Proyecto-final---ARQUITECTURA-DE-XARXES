```python
from scapy.all import *

# Crear pregunta DNS para saber la IP de google.com
paquete = IP(dst="8.8.8.8")/UDP(dport=53)/DNS(rd=1, qd=DNSQR(qname="google.com"))

# Enviar y esperar respuesta
respuesta = sr1(paquete, timeout=5)

if respuesta:
    print("Respuesta DNS:")
    respuesta.show()
else:
    print("No hubo respuesta")
```
